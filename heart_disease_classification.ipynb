```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
model=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000))
model.fit(x_train,y_train)
```

## Cross-Validation Results

We used **5-Fold Cross-Validation** to evaluate the stability of our Logistic Regression model across different data splits.

| Fold    Accuracy |

| Fold 1 -> 81.63% |
| Fold 2 -> 85.42% |
| Fold 3 -> 75.00% |
| Fold 4 -> 87.50% |
| Fold 5 -> 81.25% |
Mean Accuracy => 82.16%

### Interpretation

The model achieved a mean cross-validation accuracy of **82.16%**. The scores ranged from **75.00% to 87.50%**, indicating some variation in performance across the five folds.

This suggests that the model's performance is not fully consistent across these data splits. Further evaluation may help identify opportunities for improvement.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier 
tree_model=make_pipeline(StandardScaler(),DecisionTreeClassifier(random_state=42))
```

```python
from sklearn.model_selection import StratifiedKFold,cross_validate
```

```python
cv=StratifiedKFold(n_splits=5,
                  shuffle=True,
                  random_state=42
                  )
```

```python
scoring={
    "accuracy":"accuracy",
    "precision":"precision",
    "recall":"recall",
    "f1":"f1"
}
```

```python
log_scores=cross_validate(model,
                          x_train,
                          y_train,
                          cv=cv,
                          scoring=scoring)
```

```python
tree_scores=cross_validate(tree_model,
                           x_train,
                           y_train,
                           cv=cv,
                           scoring=scoring)
```

```python
for name, scores in [
    ("Logistic Regression", log_scores),
    ("Decision Tree", tree_scores)
]:
    print("\n", name)

    for metric in scoring:
        values = scores[f"test_{metric}"]
        print(f"{metric}: {values.mean():.3f}")
        print(f"Standard Deviation: {values.std():.3f}")
```

```python
from sklearn.metrics import accuracy_score
y_pred=model.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
f" test accuracy:{accuracy :.2%}"
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

cm
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
